**District-wise Annual Data on Area, Yield, and Production**

This dataset is organized into two main files:

* **Area, Production, and Yield:**
      Contains annual data for 20 major crops, including cereals, pulses, oilseeds, cotton, sugarcane, and aggregated totals for fruits and vegetables.       Yield figures are derived by dividing total production by the cultivated area.

* **High Yielding Varieties:**
Provides information on the adoption and distribution of high-yielding crop varieties.

The dataset captures the yearly area and production statistics for each crop. The percentage share of each crop’s area is calculated relative to the Gross Cropped Area (GCA), which is computed using a standardized methodology.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def plot_top_rice_production(file_path):
    try:
        # Load the CSV file
        df = pd.read_csv(file_path)

        # Check if required columns exist
        required_columns = ['State Name', 'RICE PRODUCTION (1000 tons)']
        for col in required_columns:
            if col not in df.columns:
                raise KeyError(f"Missing required column: '{col}' in the dataset.")

        # Group by 'State Name' and sum 'RICE PRODUCTION (1000 tons)'
        rice = df.groupby('State Name')['RICE PRODUCTION (1000 tons)'].sum()

        # Sort and select top 7 states
        top_rice = rice.sort_values(ascending=False).head(7)
        top_rice = pd.DataFrame(top_rice).reset_index()

        # Plotting
        plt.figure(figsize=(12, 6))
        sns.barplot(x='State Name', y='RICE PRODUCTION (1000 tons)', data=top_rice, palette="bright")
        plt.title('Top 7 RICE PRODUCTION State Data', fontsize=18, weight='bold')
        plt.xlabel('State Names', fontsize=15, weight='bold')
        plt.ylabel('Total_productions', fontsize=15, weight='bold')
        plt.xticks(rotation=45, ha='right', fontsize=9, weight='bold')

        # Add data labels
        for i in range(len(top_rice)):
            plt.text(i, top_rice['RICE PRODUCTION (1000 tons)'].iloc[i],
                     f"{top_rice['RICE PRODUCTION (1000 tons)'].iloc[i]:.2f}",
                     ha='center', fontsize=12, fontweight='bold')

        # Add grid lines
        plt.grid(axis='y', linestyle='--')
        plt.tight_layout()
        plt.show()

    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")
    except KeyError as e:
        print(f"Error: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# Run the function with the provided file
plot_top_rice_production("ICRISAT-District Level Data.csv")
```

1. Feature Importance Visualization (Random Forest)

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv("ICRISAT-District Level Data.csv")
df = df.select_dtypes(include=[np.number]).dropna()

# Target and features
target = 'RICE YIELD (Kg per ha)'
X = df.drop(columns=[target])
y = df[target]

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

```python
# Train Random Forest
rf_model = RandomForestRegressor(n_estimators=100, max_depth=10, random_state=42)
rf_model.fit(X_train, y_train)

# Feature importance plot
importances = rf_model.feature_importances_
feature_names = X.columns

plt.figure(figsize=(12, 6))
plt.barh(feature_names, importances)
plt.xlabel("Feature Importance")
plt.title("Random Forest Feature Importances for Rice Yield")
plt.tight_layout()
plt.show()
```

2. Model Comparison

```python
models = {
    "Linear Regression": LinearRegression(),
    "Random Forest": rf_model,
    "Gradient Boosting": GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, max_depth=5, random_state=42)
}

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    print(f"{name} - MSE: {mse:.2f}, R²: {r2:.2f}")
```

Predicting Other Crop Yields (Maize)

```python
target_maize = 'MAIZE YIELD (Kg per ha)'
X_maize = df.drop(columns=[target_maize])
y_maize = df[target_maize]

X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(scaler.fit_transform(X_maize), y_maize, test_size=0.2, random_state=42)

maize_model = RandomForestRegressor(n_estimators=100, max_depth=10, random_state=42)
maize_model.fit(X_train_m, y_train_m)
y_pred_m = maize_model.predict(X_test_m)

print("Maize Yield Prediction")
print(f"MSE: {mean_squared_error(y_test_m, y_pred_m):.2f}")
print(f"R²: {r2_score(y_test_m, y_pred_m):.2f}")
```

4. Time-Series Trend (Rice Yield Over Time)

```python
df_full = pd.read_csv("ICRISAT-District Level Data.csv")
rice_trend = df_full.groupby("Year")["RICE YIELD (Kg per ha)"].mean()

plt.figure(figsize=(10, 5))
plt.plot(rice_trend.index, rice_trend.values, marker='o')
plt.title("Average Rice Yield Over Time")
plt.xlabel("Year")
plt.ylabel("Yield (Kg per ha)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

1. Evaluation Metrics for Regression Models

Since we're predicting continuous values (e.g., crop yield), classification metrics like accuracy, precision, recall, and F1-score are not applicable. Instead, we use:

Mean Squared Error (MSE): Measures average squared difference between predicted and actual values.
Root Mean Squared Error (RMSE): Square root of MSE, interpretable in original units.
Mean Absolute Error (MAE): Average absolute difference.
R-squared (R²): Proportion of variance explained by the model.

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

def evaluate_model(model, X_test, y_test, name="Model"):
    y_pred = model.predict(X_test)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    
    print(f"{name} Evaluation:")
    print(f"  MSE  : {mse:.2f}")
    print(f"  RMSE : {rmse:.2f}")
    print(f"  MAE  : {mae:.2f}")
    print(f"  R²   : {r2:.2f}")
    print("-" * 40)

# Example usage:
evaluate_model(rf_model, X_test, y_test, name="Random Forest")
```

2. Cross-Validation for Robustness

```python
from sklearn.model_selection import cross_val_score

# 5-Fold Cross-Validation using R²
cv_scores = cross_val_score(rf_model, X_scaled, y, cv=5, scoring='r2')

print("Cross-Validation R² Scores:", cv_scores)
print("Average CV R² Score:", np.mean(cv_scores))
```

 Summary of Validation Techniques


Technique and Purpose:
Train/Test Split: Quick performance check on unseen data

Cross-Validation: Robustness across multiple data partitions

R² Score: Variance explained by the model

RMSE/MAE: Error magnitude in original units
